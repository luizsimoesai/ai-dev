# Interfaces

Em programação orientada a objetos, interfaces são como contratos: elas dizem quais métodos uma classe deve ter, mas não implementam esses métodos. A classe que "assina" esse contrato (ou seja, implementa a interface) é que vai escrever o que cada método faz. Metaclasses.

Diferente de linguagens como Java ou C#, o Python não tem um tipo especial chamado interface. Em python existe herança múltipla. Mas é possível fazer algo equivalente usando:

- Classes abstratas (ABC - Abstract Base Classes)

- Módulo abc (do Python)

- "Interfaces são métodos ou atributos públicos que outros objetos possam usar para se comunicar com outros objetos" - Dusty Phillips
- "O subconjunto de métodos públicos de um objeto que lhe permitem desempenhar um papel específico de um sistema" - Ramalho

```python
from abc import ABC, abstractmethod

# Isso é a "interface"
class Animal(ABC):
    @abstractmethod
    def fazer_som(self):
        pass

# Essa classe "implementa a interface"
class Cachorro(Animal):
    def fazer_som(self):
        print("Au au!")

class Gato(Animal):
    def fazer_som(self):
        print("Miau!")
```

```python
def emitir_som(animal: Animal):
    animal.fazer_som()

emitir_som(Cachorro())  # Saída: Au au!
emitir_som(Gato())      # Saída: Miau!
```

### O que é duck typing?
Duck typing é um conceito do Python (e outras linguagens dinâmicas) que diz:

"Se parece um pato, nada como um pato e grasna como um pato... então é um pato."

Ou seja: não importa o tipo do objeto, o que importa é o que ele faz.

No contexto de orientação a objetos:
Você não precisa usar interfaces ou heranças para garantir que um objeto tenha certos métodos.

Você simplesmente usa o objeto — se ele tiver o método esperado, funciona.

```python
class Pato:
    def quack(self):
        print("Quack!")

class Pessoa:
    def quack(self):
        print("Estou imitando um pato!")

def fazer_quack(ser):
    ser.quack()  # Não importa o tipo, só precisa ter o método `quack`

fazer_quack(Pato())    # Saída: Quack!
fazer_quack(Pessoa())  # Saída: Estou imitando um pato!
```

Duck typing = usar objetos pelo que fazem, não pelo que são.

Isso deixa o Python flexível e com menos burocracia, mas exige que você tenha cuidado ao usar objetos corretamente.

#### Segregação de Interfaces (Protocolos)

*Em python, protocolos e interfaces são sinônimos*

#### Protocolos

Na biblioteca padrão existem diversos exemplos de protocolos, em:
- collections.abc
- numbers

Você pode se basear neles para entender comportamentos 'pré definidos'

```python
# Exemplo

from collections.abc import Iterable

def é_iterável(obj):
    return isinstance(obj, Iterable)

# Testando:
print(é_iterável([1, 2, 3]))     # True (lista)
print(é_iterável("texto"))       # True (string)
print(é_iterável(123))           # False (int não é iterável)
```

```python
# Exemplo

from collections.abc import Mapping

class MeuDicionario(Mapping):
    def __init__(self, dados):
        self._dados = dados

    def __getitem__(self, chave):
        return self._dados[chave]

    def __iter__(self):
        return iter(self._dados)

    def __len__(self):
        return len(self._dados)

# Usando:
meu_dict = MeuDicionario({'a': 1, 'b': 2})
print(meu_dict['a'])      # Saída: 1
print(list(meu_dict))     # Saída: ['a', 'b']
```
