## Classes

## Atributos

```python
""" 
Atributos
Atributos são “Adjetivos”. A aquilo que é atribuído a algo. Ou seja...
“O carro preto” -> preto é um adjetivo referente a cor do carro.
Nesse sentido “cor” é um atributo.

“Atributos são variáveis internas de uma abstração de dados”
"""

class Carro:
    cor = "preto" # atributo

print(Carro.cor)
```

## Métodos

```python
""" 
Métodos
Métodos são a abstração de processos de uma abstração de dados.
São abstrações de processos embutidas nos dados.

“São funções internas da classe”
"""

class Carro:
    cor = "preto" # atributo
    def buzinar(): # método
        print("beeeee!!!")

Carro.buzinar()
```

```python
""" 
Métodos também podem manipular o estado dos atributos. 
O uso disso acontece usando duas palavras específicas 'self' e 'cls'.
'Quando o dado quer conversar com ele mesmo'. Ou seja myself, do inglês “Eu mesmo”.
"""

class Passaro:
    estado = 'indefinido' # atributo

    def voar(self): # método
        self.estado = 'voando'
        print(self.estado)

    def pousar(self): # método
        self.estado = 'parado'
        print(self.estado)

print(Passaro.estado) # imprime indefinido
```

```python
print(Passaro.voar()) # deu erro pois precisamos criar uma instância

# O erro aparece porque você tentou chamar um método de instância diretamente a partir da classe, sem fornecer uma instância como primeiro argumento (self).
```

## Instância

```python
""" 
Instância
Do inglês 'instance', quer dizer que 'exemplo'.
No exemplo dos pássaros, a classe Pássaros é uma abstração do pássaro. 
Ela é genérica. Já a instância é um pássaro.
A classe como tipo é uma generalização de todos os pássaros. 
O exemplo (instância) é um representante único daquele grupo de entidades.
"""

class Passaro:
    estado = 'indefinido'

    def voar(self):
        self.estado = 'voando'
        print(self.estado)

    def pousar(self):
        self.estado = 'parado'
        print(self.estado)

p1 = Passaro() # instância 1
p2 = Passaro() # isntância 2

print("Pássaro 1:")
print(p1.estado)
p1.voar()
print(p1.estado)

print(" \n")

print("Pássaro 2:")
print(p2.estado)
```

## self x cls

```python
""" 
self x cls
Usamos 'self' sempre que queremos falar com o exemplo (instância)
e
Usamos 'cls' quando queremos falar com a classe. 
Manipular variáveis da classe.
"""


class Fila:
    """
    Abstração de qualquer tipo de fila:
    - supermercado
    - processos
    ...
    """
    
    # --- 'cls' - manipulação da abstração do dado ---

    # 'c_fila' é um -> atributo da classe
    # característica comum que será mantida, ainda se alterada, junto com os exemplos (instâncias)
    c_fila = [] # c_ , exemplo referente à classe   

    # o decorador @classmethod deixa explícito que é um método de classe
    @classmethod
    def c_entrar(cls, obj): # recebe 'cls', pois a referência é da própria classe
        cls.c_fila.append(obj)
        print(cls.c_fila)

    # --- 'self' - manipulação do exemplo, da instância, representada pelo tipo de dado ---
    
    # método inicializador da classe (não é construtor)
    # inicia o exemplo (a instância). Com isso podemos tratar dinamicamente os atributos das instâncias.
    def __init__(self): # trabalha com 'self' pois está trabalhando com o exemplo da classe
        self.s_fila = []

    def s_entrar(self, obj): # s_, exemplo referente à self
        self.s_fila.append(obj)
        print(self.s_fila)
```

```python
# manipulando atributos da instância, que nascerá e morrerá com ela.

supermercado = Fila()

supermercado.s_entrar('Luiz')
supermercado.s_entrar('Graziela')
```

```python
processos = Fila()

print(processos.c_fila) # fila da Classe
print(processos.s_fila) # fila da instância
```

```python
Fila.c_entrar('Beatriz')
Fila.c_entrar('Mariana')
```

```python
# c_fila é a fila da classe (não importa a instância)

print(supermercado.c_fila)
print(processos.c_fila)
```

```python
# fila das instâncias

print(supermercado.s_fila)
print(processos.s_fila)
```

### Métodos
- Métodos de instância
    - só funcionam com a classe instanciada
    - manipulam atributos da instância
- Métodos de Classe
    - funcionam a todo momento, até mesmo na instância
    - manipulam atributos de classes
- Métodos estáticos
    - funcionam a todo momento
    - não interagem com atributos
- Métodos abstratos
    - dizem a subclasse o que ela deve implementar

```python
class Pizza:
    pedacos = 8

    def __init__(self, sabor):
        self.sabor = sabor

    def pegar_pedaco(self):
        # Método de instância
        self.pedacos -= 1

    @classmethod
    def mudar_tamanho(cls, pedacos):
        cls.pedacos = pedacos

    @staticmethod # ---> método que não interaje com nada que é da classe. Uma função simples 'jogada' dentro da classe
    def ingredientes():
        return 'molho de tomate, queijo, orégano'
```

```python
mussarela = Pizza('Mussarela')
```

```python
print(mussarela.sabor) # atributo da instância
print(mussarela.pedacos) # atributo da classe

# 'pegando pedaços da pizza'
mussarela.pegar_pedaco()
print(mussarela.pedacos)

mussarela.pegar_pedaco()
print(mussarela.pedacos)
```

```python
# observando que o atributo da classe não foi manipulado. Permanece com 8 pedaços
# uma pizza, por definição, geralmente tem 8 pedaços.
Pizza.pedacos
```

```python
# mudar o tamanho de todas as pizzas. 
# Pronto, agora as pizzas têm 12 pedaços

Pizza.mudar_tamanho(12)
```

```python
Pizza.pedacos
```

```python
# Método Estático
# Não interage nem com a classe nem com a instância

Pizza.ingredientes()
```

```python
# Outro exemplo

class MinhaClasse:
    @classmethod
    def metodo_de_classe(*args):
        print(args)

    def metodo_de_instancia(*args):
        print(args)
```

```python
c = MinhaClasse()
```

```python
c.metodo_de_classe()
```

```python
c.metodo_de_instancia() # retorna o objeto, a instância de class
```
