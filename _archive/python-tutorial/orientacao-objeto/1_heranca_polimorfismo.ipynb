# Herança

#### Exemplo

> - Número --> Inteiro 
> - Número --> Float
> - Número --> Complex

Número (Superclass) --> Inteiro (Subclass)


```python
from numbers import Number

issubclass(int, Number)
issubclass(float, Number)
issubclass(complex, Number)
```

Herança é um conceito fundamental da programação orientada a objetos que tem como objetivo resolver principalmente dois problemas:

Reutilização de abstrações de tipos de dados — evitando a duplicação de código ao permitir que classes compartilhem comportamentos comuns.

Organização hierárquica dos tipos — onde diferentes classes podem ser estruturadas em uma hierarquia, mantendo independência e especialização.

Por meio da herança, uma nova classe (chamada de subclasse) pode derivar de uma classe existente (a superclasse), herdando suas características e comportamentos (atributos e métodos). Essa subclasse pode então estender ou modificar funcionalidades específicas, sem a necessidade de alterar a definição original da superclasse. Isso promove o reaproveitamento de código, facilita a manutenção e reforça os princípios de modularidade e encapsulamento.

```python
# Herança

class Pizza:
    pedacos = 8

    @classmethod
    def mudar_tamanho(cls, novo_tamanho):
        cls.pedacos = novo_tamanho


class PizzaPequena(Pizza):
    pass

class PizzaMussarela(Pizza):
    pass

class PizzaCalabresa(Pizza):
    pass


# -- Herança Múltipla --

# class MeioAMeio(Pizza, PizzaMussarela, PizzaCalabresa): 
#    pass

# A classe Pizza já é heredada por associação:
class MeioAMeio(PizzaMussarela, PizzaCalabresa): 
    pass
```

```python
m = MeioAMeio()

isinstance(m, Pizza)
isinstance(m, PizzaMussarela)
isinstance(m, PizzaCalabresa)
```

# Polimorfismo

Polimorfismo é um dos pilares da programação orientada a objetos e significa, literalmente, "muitas formas". Ele permite que objetos de diferentes classes sejam tratados de maneira uniforme, desde que compartilhem uma interface ou uma estrutura comum (como uma superclasse ou métodos com o mesmo nome).

Para que serve:
O polimorfismo permite escrever código mais genérico e flexível, reduzindo a necessidade de verificações de tipo e facilitando a extensão do sistema. Ele torna possível, por exemplo, chamar o mesmo método em objetos de classes diferentes e obter comportamentos distintos, conforme a implementação específica de cada classe.

O nome dado a esse tipo de comportamento, "sobrescrever" um método de uma clase, é Polimorfismo

```python
# Polimorfismo

class Pizza:
    @staticmethod
    def ingredientes():
        return 'Ingredientes da Pizza'

# sobrescrever o comportamento da classe pai
class Mussarela(Pizza):
    @staticmethod
    def ingredientes():
        return ['queijo mussarela','molho de tomate','orégano']
```

```python
Pizza.ingredientes()
```

```python
Mussarela.ingredientes()
```
