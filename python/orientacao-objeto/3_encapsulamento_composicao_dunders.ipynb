# Composição

Composição é uma maneira de 'ligar' objetos com herança.
Se os tipos são 'filhos' (estendem o mesmo tipo) essa é a maneira de ligar classes.

```python
# A pizzaria faz uso do forno, 
# mas a pizzaria não é um tipo de forno

class Forno:
    def __init__(self):
        self.pizzas = []
        self.lenha = None
    
    def assar(self, pizza):
        if not self.lenha:
            print("Forno não possui lenha")

class Pizzaria:
    def __init__(self):
        self.forno = Forno() # Composição (classe 'Forno()')

    def pedido(self, pizza):
        self.forno.assar(pizza)
```

```python
pizza_planet = Pizzaria()

pizza_planet.forno.pizzas
```

```python
pizza_planet.pedido('Mussarela')
```

# __dunders__  - Métodos Especiais

```python
#__init__ (dunder init)

# O __init__ é o método inicializador
# Já o __new__ que é o método construtor

# parece que este é o único padrão de projeto que usa __new__
class Singleton:
    def __new__(cls):
        if not hasattr(cls, 'criado'):
            cls.criado = super().__new__(cls)
        return cls.criado
    

```

```python

# super()

# Quando estamos falando de herança, 
# a sub-classe deve chamar o __init__ da super classe,
# ou de sua classe pai.
# O super() tem a função de chamar o método definido na super classe.

class Passaro:
    def __init__(self, nome):
        self.nome = nome

class Canario(Passaro):
    def __init__(self, nome, cor):
        super().__init__(nome)
        self.cor = cor
```

```python
# __del__ 

# O dunder del é o finalizador.
# Chamado quando a instância está prestes a ser finalizado.
# O que vai acontecer antes do objeto ser de fato finalizado.

class MinhaClasse:
    def __new__(cls):
        print('-- Construindo --')
        return super().__new__(cls)
    def __init__(self):
        print('-- Inicializando --')
    def __del__(self):
        print('-- Finalizando --')
```

```python
a = MinhaClasse()
```

```python
del a # chamado automaticamente quando encerra o processo
```

```python
# __str__

# Chamado por str (objeto) e as funções internas format() e print()
# para calcular a representação de string 'informal' ou muito bem imprimível de um objeto.
# O valor de retorno deve ser um objeto de string

class Passaro:
    def __init__(self, nome):
        self.nome = nome

class Canario(Passaro):
    def __init__(self, nome, cor):
        super().__init__(nome)
        self.cor = cor

    def __str__(self):
        return f'Canario(nome="{self.nome}", cor="{self.cor}")'
```

```python
canarinho = Canario('Alemão', 'amarelo')

print(canarinho.nome)
print(canarinho.cor)
```
